# Student Grade Calculator

## Objective

To calculate student grades based on their marks using conditional statements such as `if`, `elif`, and `else`.

```python
# Student marks
marks = [78, 85, 92, 67, 74, 88, 95, 81, 69, 45]

marks
```

```python
# Calculate grades for each student

grades = []

for mark in marks:
    if mark < 0 or mark > 100:
        grade = "Invalid"
    elif mark >= 90:
        grade = "A+"
    elif mark >= 80:
        grade = "A"
    elif mark >= 70:
        grade = "B"
    elif mark >= 60:
        grade = "C"
    elif mark >= 50:
        grade = "D"
    else:
        grade = "F"

    grades.append(grade)

grades
```

```python
# Display student marks and grades

print("---------------------------------")
print("    Student Grade Calculator     ")
print("---------------------------------")

for i in range(len(marks)):
    print(f"Student {i + 1}: Marks = {marks[i]}, Grade = {grades[i]}")
```

```python
# Test invalid marks

test_marks = [-5, 105, 75]

for mark in test_marks:
    if mark < 0 or mark > 100:
        grade = "Invalid"
    elif mark >= 90:
        grade = "A+"
    elif mark >= 80:
        grade = "A"
    elif mark >= 70:
        grade = "B"
    elif mark >= 60:
        grade = "C"
    elif mark >= 50:
        grade = "D"
    else:
        grade = "F"

    print(f"Marks = {mark}, Grade = {grade}")
```

## Conclusion

The Student Grade Calculator was successfully developed using Python conditional statements.

The program assigns grades based on predefined marks ranges using `if`, `elif`, and `else` statements. It also handles invalid marks below 0 or above 100.
